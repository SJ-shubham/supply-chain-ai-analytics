```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = os.path.dirname(os.path.dirname(os.path.abspath('.'))) if os.path.basename(os.getcwd()) == 'notebooks' else os.getcwd()
if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)

from src.feature_engineering import ENGINEERED_DATASET_PATH
```

```python
df = pd.read_csv(ENGINEERED_DATASET_PATH, low_memory=False)
print(f"Loaded Engineered Dataset: {df.shape[0]:,} rows × {df.shape[1]} columns.")
```

## 1. Top 10 Countries by Gross Revenue

```python
plt.figure(figsize=(10, 5))
top_countries = df.groupby('Order Country')['Sales'].sum().nlargest(10).reset_index()
sns.barplot(data=top_countries, x='Sales', y='Order Country', hue='Order Country', palette='Blues_r', legend=False)
plt.title('Top 10 Order Countries by Total Gross Revenue ($)', fontsize=12, fontweight='bold')
plt.show()
```

## 2. Late Delivery Risk by Shipping Mode

```python
mode_perf = df.groupby('Shipping Mode')['Late_delivery_risk'].mean().reset_index()
mode_perf['Late_Pct'] = mode_perf['Late_delivery_risk'] * 100
plt.figure(figsize=(9, 5))
sns.barplot(data=mode_perf, x='Shipping Mode', y='Late_Pct', hue='Shipping Mode', palette='magma', legend=False)
plt.title('Late Delivery Rate (%) by Shipping Mode', fontsize=12, fontweight='bold')
plt.axhline(50, color='red', linestyle='--', label='50% Threshold')
plt.legend()
plt.show()
```
